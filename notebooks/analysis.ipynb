# Sales Data Analysis

This notebook loads, cleans, analyzes, and reports on the sales dataset.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / 'sales_analyzer').exists():
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from sales_analyzer.data_loader import load_sales_data, validate_columns
from sales_analyzer.data_cleaner import clean_sales_data
from sales_analyzer.analyzer import (
    basic_statistics, category_analysis, top_products,
    monthly_sales, customer_analysis
)
from sales_analyzer.reporter import generate_text_report, save_text_report

raw_data_path = project_root / 'data' / 'raw' / 'sales_data.csv.txt'
sales_data = load_sales_data(raw_data_path)
validate_columns(sales_data)
cleaned_data = clean_sales_data(sales_data)

print(f'Loaded rows: {len(sales_data):,}')
print(f'Cleaned rows: {len(cleaned_data):,}')
cleaned_data.head()
```

```python
stats = basic_statistics(cleaned_data)
categories = category_analysis(cleaned_data)
monthly = monthly_sales(cleaned_data)
customers = customer_analysis(cleaned_data)

print(f"Total sales: ${stats['total_sales']:,.2f}")
print(f"Average order value: ${stats['average_order_value']:,.2f}")
print(f"Total customers: {stats['unique_customers']:,}")
print(f"Repeat customers: {customers['repeat_customers']:,} ({customers['repeat_percentage']:.1f}%)")

print('\nSales by category:')
categories[['sales', 'percentage']]
```

```python
print('Top 5 products:')
top_five_products = top_products(cleaned_data, 5)
for number, (product, sales) in enumerate(top_five_products.items(), start=1):
    print(f'{number}. {product}: ${sales:,.2f}')

print('\nMonthly sales:')
monthly.to_frame('total_sales')
```

```python
report = generate_text_report(cleaned_data)
report_path = project_root / 'data' / 'reports' / 'sales_analysis_report.txt'
save_text_report(report, report_path)

print(report)
print(f'\nSaved report to: {report_path}')
```
